# Using Langchain (`OpenWeatherMapAPIWrapper`)

```python
from langchain_community.utilities import OpenWeatherMapAPIWrapper
import os
os.environ["OPENWEATHERMAP_API_KEY"] = weather_api

weather = OpenWeatherMapAPIWrapper()
```

```python
weather_data = weather.run("San Francisco")
print(weather_data)
```

```python
from langchain.agents import AgentType, initialize_agent, load_tools
from langchain_openai import OpenAI
```

```python
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')
```

```python
llm = OpenAI(temperature=0)
```

```python
llm.invoke("write me a hello world in Python")
```

```python
llm.invoke("What is the weather in San Francisco now?")
```

```python
llm.invoke("What did I just ask?")
```

```python
tools = load_tools(["openweathermap-api"], llm)

agent_chain = initialize_agent(
    tools=tools, llm=llm, agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION, verbose=True
)
```

```python
agent_chain.run("What's the humidity like in San Francisco now?")
```
