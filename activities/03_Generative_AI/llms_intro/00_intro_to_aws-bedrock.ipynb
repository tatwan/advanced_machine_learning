# Introduction to AWS Bedrock

## Overview

[Amazon Bedrock](https://aws.amazon.com/bedrock/) is a fully managed service that offers access to foundation models (FMs) from leading AI companies like Amazon, Anthropic, Cohere, Meta, and more through a single API.

### Key Benefits:
- **Single API Access**: Use one unified API to access multiple foundation models
- **OpenAI-Compatible API**: Bedrock now supports an OpenAI-compatible endpoint, making migration easy
- **Serverless**: No infrastructure to manage
- **Enterprise Security**: Built on AWS security and compliance

### Available Models:
- **Amazon Nova** (Lite, Pro, Premier) - Amazon's own foundation models
- **Claude** (Anthropic) - Advanced reasoning and analysis
- **Llama** (Meta) - Open-weight models
- **Titan** - Amazon's embedding and text models

## Using Bedrock with OpenAI-Compatible API

One of the most powerful features of AWS Bedrock is its **OpenAI-compatible API endpoint**. This allows you to:

- Use the familiar OpenAI Python SDK
- Easily migrate existing OpenAI-based code to Bedrock
- Switch between models with minimal code changes

### How It Works:
1. Point the OpenAI client to Bedrock's endpoint: `https://bedrock-runtime.{region}.amazonaws.com/openai/v1`
2. Use your Bedrock API key for authentication
3. Specify a Bedrock model ID (e.g., `us.amazon.nova-lite-v1:0`)

````python
from openai import OpenAI
from dotenv import load_dotenv
import os

# Load environment variables from .env file
load_dotenv()

# Get Bedrock API key from environment
api_key = os.getenv("AWS_BEARER_TOKEN_BEDROCK")

# Verify the key is loaded
if not api_key:
    raise ValueError("AWS_BEARER_TOKEN_BEDROCK not found in environment. Please set it in your .env file.")

print("✅ API key loaded successfully!")
````

````python
# Create OpenAI client pointing to Bedrock endpoint
client = OpenAI(
    base_url="https://bedrock-runtime.us-east-2.amazonaws.com/openai/v1",
    api_key=api_key
)

# Make a chat completion request using Amazon Nova Lite
completion = client.chat.completions.create(
    model="us.amazon.nova-lite-v1:0",
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant."
        },
        {
            "role": "user",
            "content": "Hello! What is AWS Bedrock?"
        }
    ]
)

# Print the response
print(completion.choices[0].message.content)
````

## Understanding the Code

Let's break down what's happening:

| Component | Description |
|-----------|-------------|
| `base_url` | Bedrock's OpenAI-compatible endpoint for your region |
| `api_key` | Your Bedrock API key (Bearer token authentication) |
| `model` | Bedrock model ID (format: `region.model-name-version`) |
| `messages` | Standard chat format with `system` and `user` roles |

### Available Nova Models:
- `us.amazon.nova-lite-v1:0` - Fast, cost-effective for simple tasks
- `us.amazon.nova-pro-v1:0` - Balanced performance and accuracy
- `us.amazon.nova-premier-v1:0` - Highest capability for complex tasks
